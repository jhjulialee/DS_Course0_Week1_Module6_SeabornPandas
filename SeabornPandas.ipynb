## Part I: Seaborn

We will use a randomly generated data set to practice using Seaborn. Begin by running the below code without change.

````python
# CodeGrade step0
# Run this cell without changes

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# The seed must be 42 for the data to replicate
seed = 42

# Data
data = np.random.normal(size=(20, 10)) + np.arange(10) / 2
````

The setup above declares `seed = 42` but does not apply it. This separate cell enables inline plots and regenerates the same simulated design with a local seeded generator, leaving the supplied setup cell unchanged.

````python
%matplotlib inline

rng = np.random.RandomState(seed)
data = rng.normal(size=(20, 10)) + np.arange(10) / 2
````

### Step 1

Create a boxplot and store the object returned in the variable boxplot1.

````python
# CodeGrade step1

fig, ax = plt.subplots(figsize=(8, 4.5))
boxplot1 = sns.boxplot(data=data, ax=ax)
plt.show()
````

### Step 2
Repeat step 1 creating another boxplot, but now also call the boxplot object's set() method in order to set the title and axis labels as follow:
* X axis should be labeled 'X Label'
* y axis should be labeled 'Y Label'
* Title should be labeled 'Example Boxplot'

````python
# CodeGrade step2

fig, ax = plt.subplots(figsize=(8, 4.5))
boxplot2 = sns.boxplot(data=data, ax=ax)
boxplot2.set(xlabel="X Label", ylabel="Y Label", title="Example Boxplot")
fig.tight_layout()
plt.show()
````

### Step 3

Repeat step 2, this time also utilizing Seaborn to set the style to be 'darkgrid'. Still include title and axis labels.

````python
# CodeGrade step3

sns.set_style("darkgrid")

fig, ax = plt.subplots(figsize=(8, 4.5))
boxplot3 = sns.boxplot(data=data, ax=ax)
boxplot3.set(xlabel="X Label", ylabel="Y Label", title="Example Boxplot")
fig.tight_layout()
plt.show()
````

### Step 4

Recreate the labeled boxplot that we made in Step 3
* Utilizing Seaborn's context setting, adjust the size and font style of text so that it is more legible for presentations and large screen format
* Use 'poster' from Seaborns preconfigured options



````python
# CodeGrade step4

sns.set_style("darkgrid")
sns.set_context("poster")

fig, ax = plt.subplots(figsize=(9, 5))
boxplot4 = sns.boxplot(data=data, ax=ax)
boxplot4.set(xlabel="X Label", ylabel="Y Label", title="Example Boxplot")
fig.tight_layout()
plt.show()
````

All four boxplots use the same simulated data. The dark grid changes the background, and the `poster` context enlarges labels without changing the distributions.

### Step 5

You are now going to take a look at the canonical Seaborn Penguins dataset. This dataset contains biometric measurements and categorical information regarding three species of penguins.

````python
# CodeGrade step0
# Run this cell without changes

penguins = pd.read_csv("penguins.csv")
penguins.head()
````

Now use Seaborn to create a histogram of the 'body_mass_g' column with the following parameters:
*   Context is set to 'talk'
*   Color of the bars should distinguish male and female from each other
*   The labels should be:
    - X Axis: 'Body Mass (g)'
    - Y Axis: 'Number of Penguins'
    - Title: 'Penguin Mass Distribution by Sex'

````python
# CodeGrade step5

sns.set_context("talk")

fig, ax = plt.subplots(figsize=(10, 5.5))
histplot1 = sns.histplot(
    data=penguins,
    x="body_mass_g",
    hue="sex",
    hue_order=["Female", "Male"],
    palette="colorblind",
    bins=15,
    common_bins=True,
    multiple="layer",
    alpha=0.45,
    ax=ax,
)
histplot1.set(
    xlabel="Body Mass (g)",
    ylabel="Number of Penguins",
    title="Penguin Mass Distribution by Sex",
)
fig.tight_layout()
plt.show()
````

The histogram includes 333 of 344 records: 165 Female and 168 Male. Records without a reported sex cannot be assigned a hue. Median body mass is 3,650 g for Female records and 4,300 g for Male records, but the distributions overlap. This comparison pools species and does not isolate a within-species difference.

### Step 6

Create a scatter plot of the bill length (horizontal axis) vs. the bill depth (vertical axis) with the following parameters:

* The context set to 'paper'
* Have the points color represent the sex
* Have the points shape/style represent the species
* The labels should be:
    - X Axis: 'Bill Length (mm)'

## Part II: Pandas

### Visualizing High Dimensional Data

You are now going to take a look at the canonical iris dataset. This dataset is a classic multivariate dataset, which includes the sepal length, sepal width, petal length, and petal width for hundreds of samples of three species of the iris flower.

````python
# CodeGrade step0
# Run this cell without changes

iris = pd.read_csv("iris.csv")
iris.head()
````

### Step 7

A primary question you might ask regarding this data is if there is any difference in average measurements across species? You can help answer this via a simple bar chart visual. In order to do so you must first create your aggregated data. You are interestred in seeing how much the mean sepal width differs across the three species.
* X Axis should be labeled as 'Species'
* Y Axis should be labeled as 'Mean Sepal Width'
* Title should be labeled as 'Distribution of Sepal Width'

````python
# CodeGrade step7

species_mean_sepal_width = iris.groupby("species")["sepal_width"].mean()

with sns.plotting_context("notebook"):
    fig, ax = plt.subplots(figsize=(7, 4))
    barplot1 = species_mean_sepal_width.plot(
        kind="bar",
        title="Distribution of Sepal Width",
        xlabel="Species",
        ylabel="Mean Sepal Width",
        color="#38788A",
        rot=0,
        ax=ax,
    )
    barplot1.figure.tight_layout()
    plt.show()

species_mean_sepal_width.round(3)
````

Setosa has the largest mean sepal width (3.428 cm), followed by Virginica (2.974 cm) and Versicolor (2.770 cm). These are means from 50 flowers per species; the chart does not show variation within a species.

### Step 8

Utilize pandas plotting (.plot) to create a histogram plot to show the distribution of the 'sepal_length' column separated and grouped by 'species'. You should end up with a chart showing three histograms, one on top of the other. You can accomplish this with one .plot() call.

````python
# CodeGrade step8

# Place each species in its own column, keeping individual lengths intact.
iris_lengths = iris.loc[:, ["species", "sepal_length"]].assign(
    observation=iris.groupby("species").cumcount()
)
sepal_length_by_species = iris_lengths.pivot(
    index="observation", columns="species", values="sepal_length"
)
sepal_length_bins = np.histogram_bin_edges(iris["sepal_length"].dropna(), bins=12)

with sns.plotting_context("notebook"):
    histplot2 = sepal_length_by_species.plot(
        kind="hist",
        subplots=True,
        layout=(3, 1),
        bins=sepal_length_bins,
        sharex=True,
        sharey=True,
        legend=False,
        color="#38788A",
        edgecolor="white",
        figsize=(8, 7.5),
    )
    iris_hist_axes = np.asarray(histplot2).ravel()
    for species, axis in zip(sepal_length_by_species.columns, iris_hist_axes):
        axis.set(title=species.capitalize(), ylabel="Count")
    iris_hist_axes[-1].set_xlabel("Sepal Length (cm)")
    iris_hist_figure = iris_hist_axes[0].figure
    iris_hist_figure.suptitle("Distribution of Sepal Length by Species")
    iris_hist_figure.tight_layout(rect=(0, 0, 1, 0.95))
    plt.show()
````

The panels show the original sepal lengths, not species means. Setosa is concentrated at shorter lengths, while Versicolor and Virginica overlap. Shared bin edges and common x and y scales make this comparison meaningful.
